# **Exploring** data

```python
df_film = pd.read_csv("tmdb_5000_credits.csv")
pd.set_option('display.max_colwidth', None)
df_film.head(1)
```

# Change from json to table

```python
import ast
df_film['cast'] = df_film['cast'].apply(ast.literal_eval)
df_film['crew'] = df_film['crew'].apply(ast.literal_eval)
```

```python
type(df_film['cast'].iloc[0])
```

# changing **cast** column

```python
df_cast_exploded = df_film.explode('cast')
df_cast_clean = pd.json_normalize(df_cast_exploded['cast'])

df_cast_clean['movie_id'] = df_cast_exploded['movie_id'].values
df_cast_clean['title'] = df_cast_exploded['title'].values
df_cast_clean.head()
```

# changing **crew** column

```python
df_crew_exploded = df_film.explode('crew')
df_crew_clean = pd.json_normalize(df_crew_exploded['crew'])
df_crew_clean['movie_id'] = df_crew_exploded['movie_id'].values
df_crew_clean['title'] = df_crew_exploded['title'].values
df_crew_clean.head()
```

# 2 sheets cleaned **connected** by id

```python
with pd.ExcelWriter("tmdb_credits_cleaned.xlsx", engine='openpyxl') as writer:
    df_cast_clean.to_excel(writer, sheet_name="Cast_Details", index=False)
    df_crew_clean.to_excel(writer, sheet_name="Crew_Details", index=False)
```
